## Etapa 1 - Carregar os dados

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import numpy as np

df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/planilha/base_custos_transporte.csv')

df.head()
```

## Etapa 2 - Identificar as variáveis

a - Variável Dependente: custo_transporte. É elaa que vamos prever ou estimar. O valor dela aparentemente se altera de acordo com o mês.

b - Variável Independente: mes. É o fato que manipulamos e que influencia na outra.

c - A variável mes representa o periodo de tempo em que os custos foram registrados. Já a variável custo_transporte representa o custo do transporte naquele mês específico.



## Etapa 3 - Construindo o gráfico

```python

plt.figure(figsize=(8,5))
plt.scatter(df['mes'], df['custo_transporte'], s=60, color='#2E86AB',edgecolor = 'black')
plt.xlabel('Mês')
plt.ylabel('Custo do transporte (R$)')
plt.title('Relação entre mês e custo de transporte')
plt.xticks(df['mes'])
plt.gca().yaxis.set_major_formatter(mticker.StrMethodFormatter('R$ {x:,.0f}'))
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

Com base no gráfico podemos observar que os pontos formam um padrão crescente, demosntrado uma forte relação entre as duas variáveis apresentadas. O cresncimento aparenta ser regular e progressivo, sem valores demasiado altos ou baixos.

## Etapa 4 - Encontrar uma função



```python
meses = df['mes'].values
custos = df['custo_transporte'].values

a,b = np.polyfit(meses,custos,1)

print(f"a = {a:.2f}")
print(f"b = {b:.2f}")
print(f"Função encontrada: f(x) = {a:.2f}x + {b:.2f}")
```

## Etapa 5 – Interpretar a função

a - O coeficiente a é taxa de variação mensal do custo de trasnporte.  Ou seja, a cada mês que se passa, o custo sobe em média R$1791,61, e a sua inclinção indica que quando maior o valor de , mais cresce o valor dos custos.

b - O coeficiente b é o nosso intercepto, o valor base quando o X = 0. Ele é a nossa eestimativa de custo base de transporte.

c - O valor dee a indica crescimento, pois o valor de a é positivo, e conforme o mês aumenta, o custo também aumenta.

## Etapa 6 – Representar a função graficamente



```python
x_reta = np.linspace(0, 13, 100)
y_reta = a * x_reta + b

plt.figure(figsize=(8, 5))
plt.scatter(meses, custos, color='#2E86AB', s=60, edgecolor='black', zorder=3, label='Dados reais')
plt.plot(x_reta, y_reta, color='#E63946',  linestyle='--', linewidth=2, zorder=2, label=f'f(x) = {a:.2f}x + {b:.2f}')

plt.title('Custo de Transporte: Dados Reais x Função Ajustada', fontsize=13, fontweight='bold')
plt.xlabel('Mês')
plt.ylabel('Custo de Transporte (R$)')
plt.xticks(range(1, 13))
plt.gca().yaxis.set_major_formatter(mticker.StrMethodFormatter('R$ {x:,.0f}'))
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper left')
plt.tight_layout()
plt.savefig('grafico_funcao_ajustada.png', dpi=150)
plt.show()
```
